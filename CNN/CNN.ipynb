```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
from tensorflow.keras.datasets import mnist
```

```python
(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

```python
#number of image, height,width
print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
```

```python
#orginally the pixel reange is from 0-255 so by dividing it with 255 it become in range of 0-1 .Its for calculation
# Why?
# Neural networks generally train more conveniently when input features have a controlled numerical scale rather than large raw pixel values.
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0
```

```python
#1 explain that these images are grayscaled image because conv2d need grascaled image
# -1 automatically calculate the size of that dimension based on the total number of items in the dataset
X_train = X_train.reshape(-1, 28, 28, 1)
X_test = X_test.reshape(-1, 28, 28, 1)
```

```python
# Create 32 filters.Each filter learns to recognize some visual pattern.
#Conv2D Layer Formula: {Output} = {Input} - ({Filter Size}) + 1) (assuming a standard stride of 1)
# MaxPooling Layer Formula: {Output} = {Input}/{Pool Size} (rounded down if it is a decimal)

model = Sequential([
    Conv2D(
        32,
        kernel_size=(3, 3),
        activation="relu",
        input_shape=(28, 28, 1)
    ),

    MaxPooling2D(pool_size=(2, 2)),

    Conv2D(
        64,
        kernel_size=(3, 3),
        activation="relu"
    ),

    MaxPooling2D(pool_size=(2, 2)),

    Flatten(),

    Dense(128, activation="relu"),

    Dense(10, activation="softmax")
])
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
model.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=32,
    validation_split=0.1
)
```

```python
# How well does my trained CNN work on unseen data
test_loss, test_accuracy = model.evaluate(X_test, y_test)

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
```

```python
predictions = model.predict(X_test)
```

```python
predicted_class = tf.argmax(predictions[50]).numpy()

print("Predicted digit:", predicted_class)
print("Actual digit:", y_test[50])
```
